## Tujuan Notebook

Notebook ini berisi **template prompt work specification** untuk mahasiswa geofisika yang ingin mendalami suatu topik keilmuan (seismologi, gravitasi & isostasi, geomagnetik, geolistrik/geokimia air, geodinamika, pemodelan numerik, dsb.) dengan AI, mengikuti prinsip dari referensi:

- Prompt diperlakukan sebagai **algorithm** dengan input, constraint, dan output contract — bukan sekadar pertanyaan bebas.
- Role membantu vocabulary & prioritas, **bukan pengganti evidence** dan bukan alasan membuat klaim tanpa dasar.
- Setiap komponen yang menentukan makna data, batas pekerjaan, atau cara menerima hasil harus ditulis **eksplisit**.


## Delapan Komponen Prompt (P = ⟨R,T,C,I,K,O,Q,V⟩)

| Komponen | Simbol | Fungsi | Prinsip kunci dari referensi |
|---|---|---|---|
| Role | **R** | Sudut pandang kerja AI | Menentukan vocabulary & jenis klarifikasi; bukan bukti credential |
| Task | **T** | Action + deliverable utama | Gunakan verb yang dapat diperiksa (derive, estimate, simulate, quantify, dst.) |
| Context | **C** | Tujuan ilmiah & tahap workflow | Hanya informasi yang membantu memilih method/interpretasi |
| Input | **I** | Data/materi yang dapat diproses | Sebutkan source, unit, convention, provenance; larang input yang tidak boleh dipakai |
| Constraints | **K** | Batas pekerjaan yang dapat diuji | File yang boleh diubah, method yang dilarang, unit yang harus dipertahankan |
| Output | **O** | Bentuk & lokasi deliverable | Path, field penting, format angka/unit; jawaban chat ≠ artifact otomatis |
| Quality criteria | **Q** | Kapan output layak dipakai | Terukur (schema valid, tolerance, preservation), bukan kata umum "akurat" |
| Verification (opsional) | **V** | Pemeriksaan proporsional risiko | Bukan berarti boleh diabaikan untuk hasil publikasi/sensitif |


## Template Prompt Geofisika (Isi per Komponen)

```markdown
### ROLE (R)
Bertindaklah sebagai [ROLE GEOFISIKA, misal: "reviewer metode seismologi" /
"numerical-model implementation reviewer untuk pemodelan gravitasi" /
"scientific Python developer untuk QA/QC data geolistrik"].
Gunakan role ini hanya karena ia mengubah cara task dikerjakan (vocabulary,
prioritas, jenis pertanyaan klarifikasi) — bukan sebagai pengganti referensi
atau dasar untuk membuat klaim yang tidak didukung input.

### TASK (T)
[VERB YANG DAPAT DIPERIKSA] + [OBJECT] + [UNIT/DOMAIN] + [KONDISI PENERIMAAN].
Contoh pola verba geofisika: derive, calculate, simulate, invert, calibrate,
quantify, delineate, interpolate, classify, map.
Misal: "Turunkan hubungan isostasi Airy untuk topografi [DATA], nyatakan
asumsi densitas kerak dan mantel, dan bandingkan dengan model Pratt untuk
kasus yang sama."
Jika task memiliki beberapa hasil, pecah menjadi langkah/recipe terpisah.

### CONTEXT (C)
Jelaskan tujuan ilmiah, tahap workflow, dan audiens:
- Mata kuliah/topik: [misal: "Praktikum Seismologi — Diagram Wadati"]
- Tahap: [eksplorasi awal / tugas pendahuluan / persiapan manuscript]
- Prasyarat yang sudah dikuasai: [misal: "fisika dasar, matematika geofisika"]
- Batas relevansi: masukkan hanya info yang memengaruhi pemilihan method
  atau interpretasi hasil — hindari context yang terlalu luas.

### INPUT (I)
- Source utama: [path/dataset/rumus/buku rujukan, misal: Fowler "The Solid
  Earth"]
- Variable/field penting & unit: [misal: travel time S-P dalam detik,
  jarak episenter dalam km]
- Coordinate/time convention (jika relevan): [CRS, UTC, depth positif ke
  bawah, dst.]
- Provenance (jika ada): [sumber data, tanggal akuisisi]
- Input yang TIDAK boleh dipakai: [misal: data lain di luar yang disebut]

### CONSTRAINTS (K)
- Method yang dilarang/diwajibkan: [misal: "gunakan hanya linear regression
  untuk fitting Wadati, jangan pakai machine learning black-box"]
- Batas software/library: [misal: "Python standard library, NumPy, pandas,
  matplotlib saja"]
- Konvensi ilmiah yang harus dipertahankan: [unit, sign convention, axis
  order]
- Jika ada ambiguitas, AI harus bertanya klarifikasi — bukan menebak sendiri.
- Larangan side effect: [misal: "jangan mengubah data mentah/raw"]

### OUTPUT (O)
Bentuk deliverable yang diminta (struktur saja, TIDAK dieksekusi pada
tahap ini):
[misal: "penjelasan konsep + turunan rumus + satu contoh numerik +
interpretasi fisis", disimpan sebagai/ditampilkan dalam bentuk apa]

### QUALITY CRITERIA (Q)
- Akurasi keilmuan sesuai literatur/konsensus geofisika terkini
- Konsistensi unit & notasi (SI) di seluruh penurunan
- Asumsi dinyatakan eksplisit sebelum digunakan
- Alur logika runtut: asumsi → penurunan → kesimpulan
- Traceable ke referensi (dapat ditelusuri sumbernya)
- Memisahkan observasi/perhitungan dari inferensi & ketidakpastian

### VERIFICATION (V) — opsional, proporsional dengan risiko
Jika menyangkut rumus/nilai kuantitatif:
- Minta AI menunjukkan cek dimensi/satuan
- Bandingkan hasil dengan nilai referensi pada literatur atau solusi analitik
- Untuk hasil penting (tugas dikumpulkan/publikasi), tetap lakukan review
  manusia meskipun "optional" — bukan berarti boleh diabaikan.
```


## Contoh Prompt Siap Pakai (Template Kosong untuk Diisi)

```text
Bertindaklah sebagai [ROLE].
Tugas Anda: [TASK — verb + object + unit + kondisi penerimaan].
Konteks: [CONTEXT — tujuan ilmiah, tahap workflow, prasyarat].
Input yang saya berikan: [INPUT — source, unit, convention, provenance].
Batasan yang harus dipatuhi: [CONSTRAINTS — method, software, konvensi].
Sampaikan dalam bentuk: [OUTPUT — struktur deliverable, belum dieksekusi].
Kriteria kualitas yang harus dipenuhi: [QUALITY CRITERIA — terukur].
Jika relevan, sertakan langkah verifikasi: [VERIFICATION].
```


## Checklist Prompt yang Baik (adaptasi dari referensi)

1. Task & object spesifik (verb + object + batas scope jelas).
2. Input dapat ditelusuri (path/source, unit, convention disebutkan eksplisit).
3. Context relevan, tidak dibebani info yang tidak memengaruhi hasil.
4. Constraints dapat diuji (bisa diperiksa lewat jawaban/diff/hasil).
5. Output contract jelas (bentuk, field penting, unit/presisi).
6. Acceptance/quality criteria terukur, bukan kata umum seperti "bagus".
7. Proporsional terhadap kerumitan & risiko (tugas kecil ≠ perlu pipeline besar).
8. Menyatakan batas pengetahuan: kapan AI harus berhenti/bertanya klarifikasi.

## Pitfall Umum yang Perlu Dihindari

- Meminta "analisis" tanpa verb, object, atau deliverable yang jelas.
- Mencampur banyak task sekaligus dalam satu prompt (pecah jadi beberapa prompt/tahap).
- Tidak menyebut unit, sistem koordinat, atau konvensi waktu — bisa menghasilkan perhitungan yang "benar" secara sintaks tapi salah makna.
- Menganggap Role sebagai pengganti evidence/referensi ilmiah.
- Membiarkan constraints hanya tersirat (tulis eksplisit apa yang tidak boleh diubah/dilakukan).
- Memakai kriteria kualitas subjektif ("akurat", "bagus") tanpa ukuran.
- Mengubah jawaban AI menjadi klaim ilmiah tanpa verifikasi/pembanding.
