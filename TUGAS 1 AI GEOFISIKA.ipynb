## Delapan Komponen Prompt (Recap)

| Komponen | Simbol | Fungsi | Prinsip kunci dari referensi |
|---|---|---|---|
| Role | **R** | Sudut pandang kerja AI | Menentukan vocabulary & jenis klarifikasi; bukan bukti credential |
| Task | **T** | Action + deliverable utama | Gunakan verb yang dapat diperiksa (derive, estimate, simulate, quantify, dst.) |
| Context | **C** | Tujuan ilmiah & tahap workflow | Hanya informasi yang membantu memilih method/interpretasi |
| Input | **I** | Data/materi yang dapat diproses | Sebutkan source, unit, convention, provenance; larang input yang tidak boleh dipakai |
| Constraints | **K** | Batas pekerjaan yang dapat diuji | File yang boleh diubah, method yang dilarang, unit yang harus dipertahankan |
| Output | **O** | Bentuk & lokasi deliverable | Path, field penting, format angka/unit; jawaban chat ≠ artifact otomatis |
| Quality criteria | **Q** | Kapan output layak dipakai | Terukur (schema valid, tolerance, preservation), bukan kata umum "akurat" |
| Verification (opsional) | **V** | Pemeriksaan proporsional risiko | Bukan berarti boleh diabaikan untuk hasil publikasi/sensitif |


## 1. Metode Seismik

| Komponen | Simbol | Isi Prompt |
|---|---|---|
| Role | **R** | Reviewer metode seismologi yang memahami analisis travel-time gelombang P dan S serta penentuan parameter gempa bumi. |
| Task | **T** | Hitung waktu tiba gelombang P dan S dari data seismogram [DATA_SEISMOGRAM], tentukan selisih waktu S-P, lalu estimasi jarak episenter menggunakan diagram Wadati; nyatakan asumsi Vp, Vs yang digunakan. |
| Context | **C** | Topik: Praktikum Seismologi — Diagram Wadati. Tahap: tugas pendahuluan sebelum interpretasi. Prasyarat: dasar teori gelombang seismik. Audiens: laporan praktikum S1 Geofisika. |
| Input | **I** | File seismogram/waveform [FORMAT: SAC/MiniSEED/CSV]; waktu tiba P (t_P) dan S (t_S) dalam detik (UTC); Vp, Vs dalam km/s (sumber literatur); nama stasiun & tanggal event. Data event lain di luar yang disebut TIDAK boleh dipakai. |
| Constraints | **K** | Gunakan hanya linear regression pada diagram Wadati; jangan machine learning black-box. Software: Python (NumPy, pandas, matplotlib). Satuan detik & km konsisten. Jika data tidak lengkap, minta klarifikasi. |
| Output | **O** | Penjelasan konsep diagram Wadati + rumus t_S − t_P terhadap jarak + satu contoh numerik + grafik/plot Wadati. |
| Quality criteria | **Q** | Rumus & satuan konsisten literatur seismologi standar; asumsi Vp/Vs eksplisit dengan sumber; alur data mentah → S-P time → regresi → estimasi jarak; traceable ke referensi (Kanamori, Lay & Wallace). |
| Verification (opsional) | **V** | Bandingkan estimasi jarak episenter dengan katalog gempa (jika tersedia); cek dimensi/satuan tiap langkah. |


## 2. Metode Gravitasi

| Komponen | Simbol | Isi Prompt |
|---|---|---|
| Role | **R** | Numerical-model implementation reviewer untuk pemodelan gravitasi dan isostasi kerak bumi. |
| Task | **T** | Turunkan hubungan isostasi Airy untuk topografi [DATA_TOPOGRAFI], hitung ketebalan root kompensasi (Δh), bandingkan dengan model Pratt untuk kasus yang sama; nyatakan asumsi densitas kerak & mantel. |
| Context | **C** | Topik: Isostasi & Keseimbangan Litosfer. Tahap: penyusunan materi/analisis konsep. Prasyarat: mekanika fluida statis, densitas & tekanan hidrostatik. Audiens: presentasi mata kuliah Geofisika. |
| Input | **I** | Data elevasi topografi [PATH/DATASET]; densitas kerak (ρc), mantel (ρm), air (jika relevan) dalam kg/m^3; referensi: Fowler "The Solid Earth". Nilai densitas tanpa sumber literatur TIDAK boleh dipakai. |
| Constraints | **K** | Gunakan hanya model Airy & Pratt yang dinyatakan; jangan model lain tanpa izin. Unit densitas (kg/m^3) & panjang (m/km) konsisten. Jika parameter densitas tidak diberikan, tanyakan nilai referensi, jangan menebak. |
| Output | **O** | Penurunan rumus isostasi Airy & Pratt + tabel perbandingan hasil ketebalan root/kompensasi + interpretasi fisis perbedaan kedua model. |
| Quality criteria | **Q** | Penurunan konsisten dengan kesetimbangan tekanan hidrostatik; unit SI dipertahankan; perbedaan asumsi Airy vs Pratt eksplisit; traceable ke Fowler atau buku geofisika standar lain. |
| Verification (opsional) | **V** | Cek konsistensi dimensi pada persamaan kesetimbangan tekanan; bandingkan orde besaran root isostasi dengan nilai tipikal literatur (puluhan km). |


## 3. Metode Geomagnetik

| Komponen | Simbol | Isi Prompt |
|---|---|---|
| Role | **R** | Scientific Python developer untuk QA/QC data geomagnetik sebelum interpretasi anomali. |
| Task | **T** | Lakukan koreksi IGRF pada data medan magnet total [DATA_MAGNETIK], hitung anomali magnetik residual, identifikasi zona anomali positif/negatif signifikan. |
| Context | **C** | Topik: Survei Geomagnetik untuk eksplorasi/penelitian struktur bawah permukaan. Tahap: pengolahan data awal sebelum pemodelan. Prasyarat: konsep medan magnet bumi & komponen IGRF. |
| Input | **I** | Data medan magnet total (nT) per titik/lintasan [PATH/DATASET]; koordinat (lat, lon), waktu (UTC); nilai IGRF referensi pada lokasi/tanggal sama; alat ukur & tanggal survei. IGRF dari lokasi/tanggal berbeda TIDAK boleh dipakai. |
| Constraints | **K** | Gunakan hanya model IGRF resmi versi yang dinyatakan. Software: Python (NumPy, pandas); jangan library koreksi pihak ketiga tanpa sumber jelas. Satuan nT konsisten. Data mentah tidak diubah, hasil koreksi disimpan sebagai data baru. |
| Output | **O** | Penjelasan konsep koreksi IGRF + rumus anomali = medan terukur − IGRF + contoh numerik + deskripsi peta/plot sebaran anomali. |
| Quality criteria | **Q** | Koreksi sesuai versi model & tanggal yang dinyatakan; satuan nT konsisten; interpretasi anomali berdasarkan kontras densitas/suseptibilitas, bukan klaim litologi pasti; traceable ke IGRF resmi (NOAA/NGDC). |
| Verification (opsional) | **V** | Bandingkan nilai IGRF yang dipakai dengan kalkulator IGRF resmi untuk lokasi & tanggal yang sama. |


## 4. Metode Geolistrik (Resistivitas)

| Komponen | Simbol | Isi Prompt |
|---|---|---|
| Role | **R** | Reviewer metode geolistrik yang memahami konfigurasi elektroda dan interpretasi resistivitas semu. |
| Task | **T** | Hitung resistivitas semu dari data arus & tegangan konfigurasi Wenner/Schlumberger [DATA_VI], susun kurva sounding resistivitas terhadap kedalaman semu. |
| Context | **C** | Topik: Survei Geolistrik untuk identifikasi lapisan akuifer/batuan. Tahap: pengolahan data lapangan awal sebelum inversi 1D/2D. Prasyarat: hukum Ohm, faktor geometri konfigurasi elektroda. |
| Input | **I** | Data arus (mA) & beda potensial (mV) per spasi elektroda [PATH/DATASET]; jarak elektroda (a/AB2), faktor geometri (K); konfigurasi Wenner atau Schlumberger (nyatakan salah satu). Data dengan noise tak difilter tanpa dinyatakan TIDAK boleh dipakai. |
| Constraints | **K** | Gunakan rumus K sesuai konfigurasi yang dinyatakan, jangan dicampur. Software: Python/Excel eksplisit, bukan software inversi black-box pada tahap ini. Satuan Ω·m konsisten. Titik data anomali diberi flag, bukan dihapus otomatis. |
| Output | **O** | Penjelasan rumus resistivitas semu (ρa = K × V/I) + tabel hasil per titik ukur + kurva sounding (ρa vs AB/2) + catatan anomali data. |
| Quality criteria | **Q** | Rumus K sesuai konfigurasi elektroda; satuan Ω·m konsisten; titik anomali diberi flag dengan alasan; traceable ke teori dasar geolistrik (misal Telford et al.). |
| Verification (opsional) | **V** | Cek ulang satu titik data secara manual (hitung K dan ρa) dan bandingkan dengan hasil program. |


## 5. Metode Georadar (GPR)

| Komponen | Simbol | Isi Prompt |
|---|---|---|
| Role | **R** | Scientific Python developer untuk QA/QC data Ground Penetrating Radar (GPR) sebelum interpretasi struktur bawah permukaan dangkal. |
| Task | **T** | Konversi data traveltime radargram [DATA_GPR] menjadi kedalaman menggunakan kecepatan rambat gelombang EM (v) pada medium yang dinyatakan, identifikasi reflektor signifikan pada radargram. |
| Context | **C** | Topik: Survei GPR untuk deteksi struktur bawah permukaan dangkal. Tahap: pengolahan data awal (time-to-depth conversion) sebelum interpretasi geologi/arkeologi. Prasyarat: gelombang EM, permitivitas dielektrik. |
| Input | **I** | Data radargram (two-way travel time) [PATH/DATASET/FORMAT]; kecepatan v (m/ns) atau permitivitas relatif (εr) dengan sumber nilai; frekuensi antena, lokasi, tanggal survei. Nilai v dari medium berbeda tanpa justifikasi TIDAK boleh dipakai. |
| Constraints | **K** | Gunakan rumus konversi d = v × t / 2 secara konsisten. Software: Python (NumPy); jangan software pengolahan GPR komersial pada tahap penyusunan prompt ini. Satuan ns & m konsisten. Permitivitas tak pasti dinyatakan sebagai asumsi eksplisit dengan rentang nilai. |
| Output | **O** | Penjelasan konsep time-to-depth conversion + rumus + contoh numerik konversi satu traveltime menjadi kedalaman + catatan reflektor yang diidentifikasi (deskriptif). |
| Quality criteria | **Q** | Rumus konversi konsisten dengan teori perambatan gelombang EM; asumsi kecepatan/permitivitas eksplisit dengan rentang wajar; reflektor dibedakan observasi vs inferensi; traceable ke referensi GPR (Annan/Reynolds). |
| Verification (opsional) | **V** | Bandingkan kedalaman hasil konversi dengan data bor/observasi lapangan (ground truth) jika tersedia. |
